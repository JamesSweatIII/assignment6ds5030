```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations
np.random.seed(42)
```

```python
def events(outcomes_die):
    set_of_events = []
    for i in range(len(outcomes_die) + 1):
        for combo in combinations(outcomes_die, i):
            set_of_events.append(combo)
    return set_of_events
```

- Imagine rolling a fair single six-sided die. There are 6 outcomes, all equally likely. Derive the sample space and the space of events. What are the probabilities of the outcomes and events? 
- Consider a random variable that assigns the square root of the number of pips on the die to each outcome. Write code to simulate rolling a single six-sided die and computing the value of the random variable. Simulate 5000 rolls and plot the mass function and ECDF of the random variable.
- Imagine rolling two fair six-sided die. Consider a random variable that adds up the pips on the dice. There are 11 outcomes (2 , 3, ..., 12), but not all are equally likely. Derive the sample space and **describe** the space of events. What are the probabilities of the outcomes?

```python
# Outcomes for rolling a single fair SIX-sided die
outcomes_six_die = [1, 2, 3, 4, 5, 6]

# The set of all events when rolling the six-sided die
set_of_events_six_die = events(outcomes_six_die)

# The probability of each outcome
prob_outcome_six_die = 1 / len(outcomes_six_die)

print("\nSingle SIX-sided die outcomes:", outcomes_six_die)
print("Set of all events when rolling the six-sided die:", set_of_events_six_die)
print("Probability of each outcome:", prob_outcome_six_die)

# Simulate rolling a single six-sided die and computing the random variable (square root of the number of pips)
num_rolls = 10000
rolls = np.random.choice(outcomes_six_die, num_rolls)
random_variable = np.sqrt(rolls)

# Plot the mass function
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
values, counts = np.unique(random_variable, return_counts=True)
plt.bar(values, counts / num_rolls, width=0.1, align='center')
plt.title('Mass Function of Random Variable (Square Root of Die Roll)')
plt.xlabel('Value of Random Variable')
plt.ylabel('Probability')
plt.grid()

# Plot the ECDF
plt.subplot(1, 2, 2)
sorted_rv = np.sort(random_variable)
ecdf = np.arange(1, num_rolls + 1) / num_rolls
plt.step(sorted_rv, ecdf, where='post')
plt.title('ECDF of Random Variable (Square Root of Die Roll)')
plt.xlabel('Value of Random Variable')
plt.ylabel('ECDF')
plt.grid()
```

To bet, you must pay a dollar, but then you get payouts that depend on how many slots are in your bet
- Basic bets:
    - Red or Black slots
    - Odd or Even slots
    - A single slots, like 20
- More complex bets:
    - Split: Two adjacent slots (e.g. {1,2})
    - Square: Four adjacent slots (e.g. {1,2,4,5})
    - Street: Three slots in a row (e.g. {1,2,3})
    - Line: Six slots (e.g. {1,2,3,4,5,6})
In general, you can only bet on 1, 2, 3, 4, 6 slots, 12 slots, or 18 slots. If your bet occurs when the wheel is spun, you gain 36/K-1 where $K$ is the number of slots you bet on; if not, you lose a dollar and get -1.

- Write code to model spinning the roulette wheel, including the colors and numbers (you could make two lists of number and color and draw a random number between 0 and 37... or use a dataframe with color and number variables and sample it... or use a dict with key to number/color pairs...)
- Describe the probability space associated with the roulette wheel: Outcomes, events, probabilities (If there are 37 outcomes, there are $2^{37}= 137,438,953,472$ events, by the way)
- You wrote code to generate a spin of the roulette wheel. Now write a function that takes a basic or complex bet as an argument, and returns the result for the player (win or lose, and the payout 36/K-1 or -1)
- Simulate betting on red, betting on odd, betting on 7, a split, and a line 1000 times each.
- Compute the average values for the bets you just simulated. What are the expected average payoffs?

```python
# Outcomes for spinning a roulette wheel
outcomes_roulette = {
    0: 'Green', 1: 'Red', 2: 'Black', 3: 'Red', 4: 'Black', 5: 'Red', 6: 'Black',
    7: 'Red', 8: 'Black', 9: 'Red', 10: 'Black', 11: 'Black', 12: 'Red', 13: 'Black',
    14: 'Red', 15: 'Black', 16: 'Red', 17: 'Black', 18: 'Red', 19: 'Red', 20: 'Black',
    21: 'Red', 22: 'Black', 23: 'Red', 24: 'Black', 25: 'Red', 26: 'Black', 27: 'Red',
    28: 'Black', 29: 'Black', 30: 'Red', 31: 'Black', 32: 'Red', 33: 'Black', 34: 'Red',
    35: 'Black', 36: 'Red'
}

# The probability of each outcome
prob_outcome_roulette = 1 / len(outcomes_roulette)

print("\nRoulette wheel outcomes (number: color):", outcomes_roulette)
print("Probability of each outcome:", prob_outcome_roulette, "\n")

# Function to simulate a spin of the roulette wheel
def spin_roulette():
    number = np.random.choice(list(outcomes_roulette.keys()))
    color = outcomes_roulette[number]
    return number, color

# Function to evaluate the bet and return the payout
def evaluate_bet(bet, spin_result):
    number, color = spin_result
    if bet == 'Red' and color == 'Red':
        return 36/18 - 1 
    elif bet == 'Black' and color == 'Black':
        return 36/18 - 1  
    elif bet == 'Odd' and number % 2 == 1 and number != 0:
        return 36/18 - 1  
    elif bet == 'Even' and number % 2 == 0 and number != 0:
        return 36/18 - 1  
    elif isinstance(bet, int) and bet == number:
        return 36/1 - 1  
    elif isinstance(bet, tuple) and len(bet) == 2 and number in bet:
        return 36/2 - 1  
    elif isinstance(bet, tuple) and len(bet) == 4 and number in bet:
        return 36/4 - 1  
    elif isinstance(bet, tuple) and len(bet) == 3 and number in bet:
        return 36/3 - 1  
    elif isinstance(bet, tuple) and len(bet) == 6 and number in bet:
        return 36/6 - 1  
    else:
        return -1  # Lose

# Simulate betting on various bets 1000 times each
bets = ['Red', 'Black', 'Odd', 'Even', 7, (1, 2), (1, 2, 3), (1, 2, 3, 4, 5, 6)]
results = {bet: [] for bet in bets}
num_simulations = 100000
for bet in bets:
    for _ in range(num_simulations):
        spin_result = spin_roulette()
        payout = evaluate_bet(bet, spin_result)
        results[bet].append(payout)
        
# Compute the average values for the bets simulated
for bet in bets:
    average_payout = np.mean(results[bet])
    print(f"Average payout for betting on {bet}: {average_payout}")
```

### 5. CDF and PDF Basics

- Verify the following functions are distribution functions and compute their density functions. Plot the distribution and density.

1. $F(x) = \begin{cases}
0, & x \le 0 \\
\sqrt{x}, & 0 \le x \le 1 \\
1, & x \ge 1 
\end{cases}
$
2. $F(x) = \dfrac{1}{1+e^{-x}}$
3. For $ a < b < c$,
$
F(x) = \begin{cases}
0, & x \le 0 \\
\frac{(x-a)^2}{(b-a)(b-c)}, & a \le x \le c \\
1 - \frac{(b-x)^2}{(b-a)(b-c)}, & c < x < b \\
1, & x \ge b
\end{cases}
$
4. $ F(x) = \begin{cases}
0, & x <0 \\
1 - e^{-x}, & x>0
\end{cases}
$
5. $ F(x) = \begin{cases}
0, & x \le 0 \\
x, & 0 < x < 1\\
1, & x \ge 1
\end{cases}
$

```python
def plot_cdf_pdf(F, x_range, title):
    x = np.linspace(x_range[0], x_range[1], 1000)
    F_values = F(x)
    f_values = np.gradient(F_values, x)

    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.plot(x, F_values, label='CDF')
    plt.title(f'Cumulative Distribution Function - {title}')
    plt.xlabel('x')
    plt.ylabel('F(x)')
    plt.grid()
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(x, f_values, label='PDF', color='orange')
    plt.title(f'Probability Density Function - {title}')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid()
    plt.legend()

    plt.show()

# 1.
def F1(x):
    # Ensure sqrt is only called for x >= 0
    x = np.array(x)
    return np.where(x <= 0, 0, np.where(x <= 1, np.sqrt(np.maximum(x, 0)), 1))
plot_cdf_pdf(F1, (-1, 2), 'F1')

# 2.
def F2(x):
    return 1 / (1 + np.exp(-x))
plot_cdf_pdf(F2, (-5, 5), 'F2')

# 3.
a, b, c = 1, 4, 2
def F3(x):
    return np.where(x <= a, 0,
                    np.where(x <= c, ((x - a) ** 2) / ((b - a) * (b - c)),
                             np.where(x < b, 1 - ((b - x) ** 2) / ((b - a) * (b - c)), 1)))
plot_cdf_pdf(F3, (0, 5), 'F3')

# 4.
def F4(x):
    return np.where(x < 0, 0, 1 - np.exp(-x))
plot_cdf_pdf(F4, (-1, 5), 'F4')

# 5.
def F5(x):
    return np.where(x <= 0, 0, np.where(x < 1, x, 1))
plot_cdf_pdf(F5, (-1, 2), 'F5')
```

### 6. Some Common Distributions

For the following distributions:
- Determine the support
- Compute the density from the distribution for the logistic and exponential distributions (take a derivative)
- Plot the density and distribution for a variety of parameter values
- Take a sample of 1000 draws $(x_1, x_2, ..., x_{1000})$ from the distribution, plot a KDE and ECDF, visually compare with the theoretical pdf/cdf
- Find an example of this general type of PDF/CDF from the Metabric cancer data

You can use https://docs.scipy.org/doc/scipy/reference/stats.html to generate values for the pdf/cdf and generate samples of random variates.


- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.


```python
from scipy import stats
import seaborn as sns

# Logistic distribution
sigmas = [0.5, 1.0, 2.0]
x_logistic = np.linspace(-10, 10, 1000)

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
for sigma in sigmas:
    plt.plot(x_logistic, stats.logistic.cdf(x_logistic, loc=0, scale=sigma), label=f'σ={sigma}')
plt.title('Logistic CDFs')
plt.xlabel('x')
plt.ylabel('F(x)')
plt.grid()
plt.legend()

plt.subplot(1, 2, 2)
for sigma in sigmas:
    plt.plot(x_logistic, stats.logistic.pdf(x_logistic, loc=0, scale=sigma), label=f'σ={sigma}')
plt.title('Logistic PDFs')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()
plt.show()

# Exponential distribution
lambdas = [0.5, 1.0, 2.0]
x_exp = np.linspace(0, 10, 1000)

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
for lam in lambdas:
    plt.plot(x_exp, stats.expon.cdf(x_exp, scale=1/lam), label=f'λ={lam}')
plt.title('Exponential CDFs')
plt.xlabel('x')
plt.ylabel('F(x)')
plt.grid()
plt.legend()

plt.subplot(1, 2, 2)
for lam in lambdas:
    plt.plot(x_exp, stats.expon.pdf(x_exp, scale=1/lam), label=f'λ={lam}')
plt.title('Exponential PDFs')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()
plt.show()

# Sample 1000 draws from logistic and exponential, plot KDE and ECDF
sample_logistic = stats.logistic.rvs(loc=0, scale=1, size=1000)
sample_expon = stats.expon.rvs(scale=1, size=1000)

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.kdeplot(sample_logistic, label='KDE Logistic')
plt.plot(x_logistic, stats.logistic.pdf(x_logistic, loc=0, scale=1), label='PDF Logistic')
plt.title('Logistic PDF vs KDE')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
sorted_logistic = np.sort(sample_logistic)
ecdf_logistic = np.arange(1, 1001) / 1000
plt.step(sorted_logistic, ecdf_logistic, label='ECDF Logistic')
plt.plot(x_logistic, stats.logistic.cdf(x_logistic, loc=0, scale=1), label='CDF Logistic')
plt.title('Logistic CDF vs ECDF')
plt.xlabel('x')
plt.ylabel('F(x)')
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.kdeplot(sample_expon, label='KDE Exponential')
plt.plot(x_exp, stats.expon.pdf(x_exp, scale=1), label='PDF Exponential')
plt.title('Exponential PDF vs KDE')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
sorted_expon = np.sort(sample_expon)
ecdf_expon = np.arange(1, 1001) / 1000
plt.step(sorted_expon, ecdf_expon, label='ECDF Exponential')
plt.plot(x_exp, stats.expon.cdf(x_exp, scale=1), label='CDF Exponential')
plt.title('Exponential CDF vs ECDF')
plt.xlabel('x')
plt.ylabel('F(x)')
plt.legend()
plt.grid()
plt.show()

# Negative Binomial distribution
r, p = 5, 0.4
k_nb = np.arange(0, 30)
pmf_nb = stats.nbinom.pmf(k_nb, r, p)
cdf_nb = stats.nbinom.cdf(k_nb, r, p)

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.stem(k_nb, pmf_nb)
plt.title('Negative Binomial PMF (r=5, p=0.4)')
plt.xlabel('k')
plt.ylabel('P(X=k)')
plt.grid()

plt.subplot(1, 2, 2)
plt.step(k_nb, cdf_nb, where='mid')
plt.title('Negative Binomial CDF (r=5, p=0.4)')
plt.xlabel('k')
plt.ylabel('F(X≤k)')
plt.grid()
plt.show()

# Categorical distribution
probs = [0.1, 0.2, 0.3, 0.4]
categories = np.arange(1, len(probs)+1)
cdf_cat = np.cumsum(probs)

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.bar(categories, probs)
plt.title('Categorical PMF')
plt.xlabel('Category')
plt.ylabel('Probability')
plt.xticks(categories)
plt.grid(axis='y')

plt.subplot(1, 2, 2)
plt.step(categories, cdf_cat, where='mid')
plt.title('Categorical CDF')
plt.xlabel('Category')
plt.ylabel('Cumulative Probability')
plt.xticks(categories)
plt.grid(axis='y')
plt.show()



# - Find an example of this general type of PDF/CDF from the Metabric cancer data
import pandas as pd
# Load the Metabric dataset
metabric = pd.read_csv('/Users/VIDAL2/Desktop/1Masters/DS5030/assignments/assignment6ds5030/metabric.csv')
# Example: Age at diagnosis
ages = metabric['Age at Diagnosis'].dropna()
# Plot KDE and ECDF for age at diagnosis
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.kdeplot(ages, label='KDE Age at Diagnosis')
plt.title('KDE of Age at Diagnosis')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend()
plt.grid()
plt.subplot(1, 2, 2)
sorted_ages = np.sort(ages)
ecdf_ages = np.arange(1, len(sorted_ages) + 1) / len(sorted_ages)
plt.step(sorted_ages, ecdf_ages, label='ECDF Age at Diagnosis')
plt.title('ECDF of Age at Diagnosis')
plt.xlabel('Age')
plt.ylabel('F(x)')
plt.legend()
plt.grid()
plt.show()

# - Find an example of this general type of PDF/CDF from the Metabric cancer data
# Example: Tumor Size
tumor_sizes = metabric['Tumor Size'].dropna()
# Plot KDE and ECDF for tumor size
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.kdeplot(tumor_sizes, label='KDE Tumor Size')
plt.title('KDE of Tumor Size')
plt.xlabel('Tumor Size (mm)')
plt.ylabel('Density')
plt.legend()
plt.grid()
plt.subplot(1, 2, 2)
sorted_sizes = np.sort(tumor_sizes)
ecdf_sizes = np.arange(1, len(sorted_sizes) + 1) / len(sorted_sizes)
plt.step(sorted_sizes, ecdf_sizes, label='ECDF Tumor Size')
plt.title('ECDF of Tumor Size')
plt.xlabel('Tumor Size (mm)')
plt.ylabel('F(x)')
plt.legend()
plt.grid()
plt.show()
```
